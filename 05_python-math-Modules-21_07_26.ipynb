### 21. Count how many bits are set to 1 in a number using bitwise operations.

```python
def count_set_bits(n):
    count = 0
    while n > 0:
        n &= (n - 1) # This clears the least significant set bit
        count += 1
    return count

number_to_check = 29 # Binary: 11101
set_bits = count_set_bits(number_to_check)
print(f"The number {number_to_check} (binary: {bin(number_to_check)}) has {set_bits} set bits.")

number_to_check = 16 # Binary: 10000
set_bits = count_set_bits(number_to_check)
print(f"The number {number_to_check} (binary: {bin(number_to_check)}) has {set_bits} set bits.")
```

### 27. Count how many vowels in a string belong to a predefined vowel list using `in`.

```python
def count_vowels(s):
    vowels = "aeiouAEIOU"
    count = 0
    for char in s:
        if char in vowels:
            count += 1
    return count

input_string = "Hello World"
vowel_count = count_vowels(input_string)
print(f"The string '{input_string}' has {vowel_count} vowels.")

input_string = "Python Programming"
vowel_count = count_vowels(input_string)
print(f"The string '{input_string}' has {vowel_count} vowels.")
```

```python
import math

def display_menu():
    print("\n--- Menu-Driven Calculator ---")
    print("1. Arithmetic Operations")
    print("2. Comparison Operations")
    print("3. Logical Operations")
    print("4. Assignment Operations (demonstration)")
    print("5. Membership Operations")
    print("6. Identity Operations")
    print("7. Bitwise Operations")
    print("0. Exit")
    print("----------------------------")

def get_numbers():
    try:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        return num1, num2
    except ValueError:
        print("Invalid input. Please enter valid numbers.")
        return None, None

def get_integer_numbers():
    try:
        num1 = int(input("Enter first integer: "))
        num2 = int(input("Enter second integer: "))
        return num1, num2
    except ValueError:
        print("Invalid input. Please enter valid integers.")
        return None, None

def do_arithmetic():
    n1, n2 = get_numbers()
    if n1 is None or n2 is None: return

    print(f"\nNumbers: {n1}, {n2}")
    print(f"Addition ({n1} + {n2}): {n1 + n2}")
    print(f"Subtraction ({n1} - {n2}): {n1 - n2}")
    print(f"Multiplication ({n1} * {n2}): {n1 * n2}")
    if n2 != 0:
        print(f"Division ({n1} / {n2}): {n1 / n2}")
        print(f"Floor Division ({n1} // {n2}): {n1 // n2}")
        print(f"Modulus ({n1} % {n2}): {n1 % n2}")
    else:
        print("Cannot divide by zero.")
    print(f"Exponentiation ({n1} ** {n2}): {n1 ** n2}")

def do_comparison():
    n1, n2 = get_numbers()
    if n1 is None or n2 is None: return

    print(f"\nNumbers: {n1}, {n2}")
    print(f"{n1} == {n2}: {n1 == n2}")
    print(f"{n1} != {n2}: {n1 != n2}")
    print(f"{n1} > {n2}: {n1 > n2}")
    print(f"{n1} < {n2}: {n1 < n2}")
    print(f"{n1} >= {n2}: {n1 >= n2}")
    print(f"{n1} <= {n2}: {n1 <= n2}")

def do_logical():
    print("\n--- Logical Operations (True/False values) ---")
    val1_str = input("Enter first boolean value (True/False): ").lower()
    val2_str = input("Enter second boolean value (True/False): ").lower()

    val1 = val1_str == 'true'
    val2 = val2_str == 'true'

    print(f"Value 1: {val1}, Value 2: {val2}")
    print(f"{val1} and {val2}: {val1 and val2}")
    print(f"{val1} or {val2}: {val1 or val2}")
    print(f"not {val1}: {not val1}")

def do_assignment_demonstration():
    print("\n--- Assignment Operators Demonstration ---")
    x = 10
    print(f"Initial value of x: {x}")
    x += 5; print(f"x += 5: {x}")
    x -= 3; print(f"x -= 3: {x}")
    x *= 2; print(f"x *= 2: {x}")
    x /= 4; print(f"x /= 4: {x}")
    x //= 2; print(f"x //= 2: {x}")
    x %= 3; print(f"x %= 3: {x}")
    x **= 2; print(f"x **= 2: {x}")

def do_membership():
    print("\n--- Membership Operators ---")
    my_list = [10, 20, 30, 40, 50]
    my_string = "Hello World"

    item = input("Enter an item to check in list (e.g., 20 or 99): ")
    char = input("Enter a character to check in string (e.g., 'o' or 'z'): ")

    try:
        item_num = int(item)
        print(f"{item_num} in {my_list}: {item_num in my_list}")
        print(f"{item_num} not in {my_list}: {item_num not in my_list}")
    except ValueError:
        print("Invalid input for list item, checking as string.")
        print(f"'{item}' in {my_list} (as string): {'item' in str(my_list)}") # Fallback to string check for non-numeric input

    print(f"'{char}' in '{my_string}': {'char' in my_string}")
    print(f"'{char}' not in '{my_string}': {'char' not in my_string}")

def do_identity():
    print("\n--- Identity Operators ---")
    a = [1, 2, 3]
    b = a
    c = [1, 2, 3]

    print(f"a: {a}, b: {b}, c: {c}")
    print(f"a is b: {a is b}")
    print(f"a is c: {a is c}")
    print(f"a is not c: {a is not c}")

    x = 5
    y = 5
    z = 6
    print(f"\nx: {x}, y: {y}, z: {z}")
    print(f"x is y: {x is y}") # Often True for small integers due to interning
    print(f"x is z: {x is z}")

def do_bitwise():
    n1, n2 = get_integer_numbers()
    if n1 is None or n2 is None: return

    print(f"\nNumbers (binary): {n1}={bin(n1)}, {n2}={bin(n2)}")
    print(f"AND ({n1} & {n2}): {n1 & n2} ({bin(n1 & n2)})")
    print(f"OR ({n1} | {n2}): {n1 | n2} ({bin(n1 | n2)})")
    print(f"XOR ({n1} ^ {n2}): {n1 ^ n2} ({bin(n1 ^ n2)})")
    print(f"Left Shift ({n1} << 1): {n1 << 1} ({bin(n1 << 1)})")
    print(f"Right Shift ({n1} >> 1): {n1 >> 1} ({bin(n1 >> 1)})")
    print(f"NOT (~{n1}): {~n1} ({bin(~n1)}) - (Python's two's complement representation)")

# Main calculator loop
while True:
    display_menu()
    choice = input("Enter your choice: ")

    if choice == '1':
        do_arithmetic()
    elif choice == '2':
        do_comparison()
    elif choice == '3':
        do_logical()
    elif choice == '4':
        do_assignment_demonstration()
    elif choice == '5':
        do_membership()
    elif choice == '6':
        do_identity()
    elif choice == '7':
        do_bitwise()
    elif choice == '0':
        print("Exiting calculator. Goodbye!")
        break
    else:
        print("Invalid choice. Please try again.")
```
