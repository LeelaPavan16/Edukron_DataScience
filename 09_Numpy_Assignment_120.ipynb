### Question 40: Low-Margin Transactions

Management considers a profit margin below 22% to be weak.

Find all store-category combinations where the profit margin is below 22%.

Display:

* Store
* Category
* Sales
* Cost
* Profit
* Profit margin

Use Boolean indexing.

````python
profit_data = sales_data - cost_data

profit_margin = (profit_data / sales_data) * 100

low_margin = profit_margin < 22

rows, cols = np.where(low_margin)

for r, c in zip(rows, cols):
    print("Store:", stores[r])
    print("Category:", categories[c])
    print("Sales:", sales_data[r, c])
    print("Cost:", cost_data[r, c])
    print("Profit:", profit_data[r, c])
    print("Profit Margin:", round(profit_margin[r, c], 2), "%")
    print("-" * 30)

profit_data = sales_data - cost_data
profit_margin = (profit_data / sales_data) * 100

low_margin = profit_margin < 22

print(low_margin)
print(np.any(low_margin))
````
