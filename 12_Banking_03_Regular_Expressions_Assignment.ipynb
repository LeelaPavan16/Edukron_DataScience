### Q31. Parse a receipt line into named groups for banking case 1; use sample reference `ACC-1031` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re

receipt = "Receipt: TXN123 | Amount: Rs.2500 | Date: 08-09-2026"

pattern = (
    r"Receipt:\s*(?P<transaction>\w+)\s*\|\s*"
    r"Amount:\s*Rs\.(?P<amount>\d+)\s*\|\s*"
    r"Date:\s*(?P<date>\d{2}-\d{2}-\d{4})"
)

match = re.fullmatch(pattern, receipt)

if match:
    data = match.groupdict()

    print("Transaction ID:", data["transaction"])
    print("Amount:", data["amount"])
    print("Date:", data["date"])
else:
    print("Invalid receipt format")
```

### Q34. Normalize account holder phone numbers for banking case 1; use sample reference `ACC-1034` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re

phone_numbers = [
    "9876543210",
    "987-654-3210",
    "987 654 3210",
    "987-6543210"
]

pattern = r"^(?P<first>[6-9]\d{2})[-\s]?(?P<middle>\d{3})[-\s]?(?P<last>\d{4})$"

for phone in phone_numbers:
    match = re.fullmatch(pattern, phone)

    if match:
        data = match.groupdict()

        normalized = (
            data["first"] +
            data["middle"] +
            data["last"]
        )

        print(phone, "->", normalized)
    else:
        print(phone, "-> Invalid")
```
