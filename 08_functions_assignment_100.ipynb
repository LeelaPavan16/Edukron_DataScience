### Assignment 13: Factorial Using a Function
`calculate_factorial(n)` using a loop (no `math.factorial`), handles negative input.

```python
def calculate_factorial(n):
    """Return factorial of n using a loop."""
    if n < 0:
        return "Error: factorial not defined for negative numbers."
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

for n in [5, 0, -3]:
    print(f"Factorial of {n}: {calculate_factorial(n)}")
```

### Assignment 15: Count Vowels in a String

```python
def count_vowels(text):
    """Return total number of vowels in text."""
    vowels = "aeiouAEIOU"
    count = 0
    for ch in text:
        if ch in vowels:
            count += 1
    return count

for text in ["Python Functions", "Hello World"]:
    print(f"'{text}' -> Number of vowels: {count_vowels(text)}")
```

### Assignment 16: Reverse a String

```python
def reverse_string_slicing(text):
    return text[::-1]

def reverse_string_loop(text):
    reversed_text = ""
    for ch in text:
        reversed_text = ch + reversed_text
    return reversed_text

for text in ["Python", "Colab"]:
    print(f"'{text}' -> Slicing: {reverse_string_slicing(text)}, Loop: {reverse_string_loop(text)}")
```

### Assignment 17: Palindrome Checker

```python
def is_palindrome(text):
    """Return True if text is a palindrome, ignoring spaces and case."""
    cleaned = text.replace(" ", "").lower()
    return cleaned == cleaned[::-1]

for text in ["Never odd or even", "Hello there"]:
    result = is_palindrome(text)
    print(f"'{text}' -> {'The text is a palindrome.' if result else 'The text is not a palindrome.'}")
```

### Assignment 19: Find Maximum in a List
`find_list_maximum(numbers)` without `max()`, handles empty list.

```python
def find_list_maximum(numbers):
    """Return largest element in list without using max()."""
    if not numbers:
        return "Error: list is empty."
    largest = numbers[0]
    for n in numbers:
        if n > largest:
            largest = n
    return largest

for numbers in [[3, 9, 1, 7], []]:
    print(f"{numbers} -> Maximum: {find_list_maximum(numbers)}")
```

### Assignment 20: Count Positive, Negative and Zero Values
`count_number_types(numbers)` returns three counts, unpacked via tuple.

```python
def count_number_types(numbers):
    """Return counts of positive, negative, and zero values."""
    positive = negative = zero = 0
    for n in numbers:
        if n > 0:
            positive += 1
        elif n < 0:
            negative += 1
        else:
            zero += 1
    return positive, negative, zero

numbers = [5, -3, 0, 8, -1, 0, 2, -7, 4, 0]
pos, neg, zer = count_number_types(numbers)
print(f"Positive numbers: {pos}")
print(f"Negative numbers: {neg}")
print(f"Zeros: {zer}")
```

### Assignment 32: Function Returning a Dictionary
`analyze_text(text)` returns character/word/vowel/digit/space counts.

```python
def analyze_text(text):
    vowels = "aeiouAEIOU"
    stats = {
        "total_characters": len(text),
        "total_words": len(text.split()),
        "total_vowels": sum(1 for ch in text if ch in vowels),
        "total_digits": sum(1 for ch in text if ch.isdigit()),
        "total_spaces": sum(1 for ch in text if ch == " "),
    }
    return stats

result = analyze_text("Python 3 is fun in 2026")
for key, value in result.items():
    print(f"{key}: {value}")
```

### Assignment 34: Prime Number Function
`is_prime(number)` used to print all primes from 1 to 100.

```python
def is_prime(number):
    """Return True if number has exactly two factors (1 and itself)."""
    if number < 2:
        return False
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False
    return True

primes = [n for n in range(1, 101) if is_prime(n)]
print("Prime numbers from 1 to 100:")
print(primes)
```

### Assignment 35: Fibonacci Function
`generate_fibonacci(terms)` returns the series as a list, handles zero/negative, no recursion.

```python
def generate_fibonacci(terms):
    if terms <= 0:
        return "Error: number of terms must be positive."
    series = []
    a, b = 0, 1
    for _ in range(terms):
        series.append(a)
        a, b = b, a + b
    return series

print(generate_fibonacci(7))
print(generate_fibonacci(0))
```

# Assignment 56: Mask Email Addresses
Given a list of email addresses, mask the username portion.

```python
def mask_email(email):
    if "@" not in email:
        return "Invalid"

    username, domain = email.split("@", 1)

    if len(username) <= 1:
        masked = username
    else:
        masked = username[0] + "*" * (len(username) - 1)

    return masked + "@" + domain

emails = input("Enter emails separated by space: ").split()

masked_emails = list(map(mask_email, emails))

print("Masked Emails:")
for original, masked in zip(emails, masked_emails):
    print(f"{original} -> {masked}")
```

# Assignment 81: Calculate Factorial Using reduce()
Write a function that calculates factorial using reduce().


```python
from functools import reduce

def factorial(n):
    if n < 0:
        return "Negative numbers are not allowed"

    if n == 0:
        return 1

    return reduce(lambda a, b: a * b, range(1, n + 1))

number = int(input("Enter a number: "))

print("Factorial:", factorial(number))
```

# Assignment 92: Closure-Based Bank Balance
Write a function named create_bank_account().

```python
def create_bank_account(initial_balance=0):
    balance = initial_balance

    def deposit(amount):
        nonlocal balance
        if amount > 0:
            balance += amount
            print("Deposited:", amount)
        else:
            print("Invalid deposit amount")

    def withdraw(amount):
        nonlocal balance
        if amount <= balance:
            balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")

    def check_balance():
        print("Current Balance:", balance)

    return deposit, withdraw, check_balance

# Create account
deposit, withdraw, check_balance = create_bank_account(1000)

check_balance()
deposit(500)
check_balance()
withdraw(300)
check_balance()
withdraw(2000)
check_balance()
```

# Assignment 93: Function Decorator
Create a decorator named execution_logger().

```python
def execution_logger(func):

    def wrapper(*args, **kwargs):
        print("Starting function...")

        result = func(*args, **kwargs)

        print("Function completed.")

        return result

    return wrapper

# Apply decorator
@execution_logger
def add(a, b):
    result = a + b
    print("Result:", result)
    return result

# Call function
final_result = add(10, 20)

print("Returned Value:", final_result)
```

```python
from functools import reduce

# -----------------------------
# Global variable (LEGB Demo)
# -----------------------------
company_name = "Edukron"

# -----------------------------
# Employee Data
# -----------------------------
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }
]

# -----------------------------
# Part 1: Display Employees
# -----------------------------
def display_employees(employee_list):
    if not employee_list:
        print("No employees found.")
        return

    for emp in employee_list:
        print("-" * 40)
        print(f"ID: {emp['id']}")
        print(f"Name: {emp['name']}")
        print(f"Department: {emp['department']}")
        print(f"Salary: {emp['salary']}")
        print(f"Performance: {emp['performance_score']}")
        print(f"Experience: {emp['experience']} years")
    print("-" * 40)

# -----------------------------
# Part 2: Extract Employee Names
# -----------------------------
def get_employee_names():
    return list(map(lambda e: e["name"], employees))

# -----------------------------
# Part 3: Filter High Performers
# -----------------------------
def get_high_performers():
    return list(filter(lambda e: e["performance_score"] >= 80, employees))

# -----------------------------
# Part 4: Filter by Department
# -----------------------------
def search_by_department(department):
    result = list(
        filter(lambda e: e["department"].lower() == department.lower(), employees)
    )
    return result

# -----------------------------
# Part 5: Salary Increase
# -----------------------------
def calculate_revised_salaries():
    def revise(emp):
        score = emp["performance_score"]

        if score >= 90:
            increase = 15
        elif score >= 80:
            increase = 10
        elif score >= 70:
            increase = 5
        else:
            increase = 0

        new_salary = emp["salary"] + (emp["salary"] * increase / 100)

        return {
            "name": emp["name"],
            "old_salary": emp["salary"],
            "increase_percent": increase,
            "new_salary": round(new_salary, 2)
        }

    return list(map(revise, employees))

# -----------------------------
# Part 6: Total Salary Expense
# -----------------------------
def salary_expense_report():
    current_total = reduce(
        lambda acc, e: acc + e["salary"],
        employees,
        0
    )

    revised = calculate_revised_salaries()

    revised_total = reduce(
        lambda acc, e: acc + e["new_salary"],
        revised,
        0
    )

    additional = revised_total - current_total

    return current_total, revised_total, round(additional, 2)

# -----------------------------
# Part 7: Highest-Paid Employee
# -----------------------------
def highest_paid_employee():
    return reduce(
        lambda a, b: a if a["salary"] >= b["salary"] else b,
        employees
    )

# -----------------------------
# Part 8: Sort Employees
# -----------------------------
def sort_employees(option):
    if option == "name":
        return sorted(employees, key=lambda e: e["name"])

    elif option == "salary":
        return sorted(employees, key=lambda e: e["salary"])

    elif option == "performance":
        return sorted(employees, key=lambda e: e["performance_score"], reverse=True)

    elif option == "experience":
        return sorted(employees, key=lambda e: e["experience"], reverse=True)

    else:
        return employees

# -----------------------------
# Part 9: Closure-Based Bonus Calculator
# -----------------------------
def create_bonus_calculator(percent):

    def calculate_bonus(salary):
        return round(salary * percent / 100, 2)

    return calculate_bonus

bonus_5 = create_bonus_calculator(5)
bonus_10 = create_bonus_calculator(10)
bonus_15 = create_bonus_calculator(15)

# -----------------------------
# Part 10 & 11: Nested Report Generator + LEGB
# -----------------------------
def generate_department_report(department_name):

    # Enclosing scope variable
    department = department_name

    dept_employees = search_by_department(department)

    if not dept_employees:
        return {"error": "Department not found"}

    # Local variable
    report_title = f"{department} Department Report"

    def employee_count():
        return len(dept_employees)

    def total_salary():
        return sum(map(lambda e: e["salary"], dept_employees))

    def average_salary():
        return round(total_salary() / employee_count(), 2)

    def average_performance():
        total = sum(map(lambda e: e["performance_score"], dept_employees))
        return round(total / employee_count(), 2)

    report = {
        "company": company_name,      # Global scope
        "department": department,     # Enclosing scope
        "title": report_title,        # Local scope
        "employee_count": employee_count(),
        "total_salary": total_salary(),
        "average_salary": average_salary(),
        "average_performance": average_performance()
    }

    return report

# -----------------------------
# Test Calls (Optional)
# -----------------------------
if __name__ == "__main__":

    print("\nEmployee Names:")
    print(get_employee_names())

    print("\nHigh Performers:")
    display_employees(get_high_performers())

    print("\nIT Department:")
    display_employees(search_by_department("IT"))

    print("\nRevised Salaries:")
    for item in calculate_revised_salaries():
        print(item)

    print("\nSalary Expense Report:")
    current, revised, extra = salary_expense_report()
    print("Current:", current)
    print("Revised:", revised)
    print("Additional:", extra)

    print("\nHighest Paid Employee:")
    print(highest_paid_employee())

    print("\nSorted by Salary:")
    display_employees(sort_employees("salary"))

    print("\nBonus on 50000:")
    print("5%:", bonus_5(50000))
    print("10%:", bonus_10(50000))
    print("15%:", bonus_15(50000))

    print("\nDepartment Report:")
    print(generate_department_report("IT"))
```
