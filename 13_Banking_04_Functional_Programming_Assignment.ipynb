### Q33. Group accounts using a helper function for banking case 1; use sample reference `ACC-1033` and explain the result.

**Hints**

1. First express the input and return value of one pure function.
2. Use `filter` for selection, `map` for transformation, and `reduce` only when a clearer built-in is not available.
3. Test the pipeline with an empty collection and avoid mutating account records in place.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
from collections import defaultdict


def get_account_category(account):
    if account["balance"] >= 50000:
        return "Premium"
    elif account["balance"] >= 20000:
        return "Standard"
    else:
        return "Basic"


def group_accounts(accounts):
    grouped = defaultdict(list)

    for account in accounts:
        category = get_account_category(account)
        grouped[category].append(account)

    return dict(grouped)


accounts = [
    {"account_no": "ACC-1031", "balance": 75000},
    {"account_no": "ACC-1032", "balance": 30000},
    {"account_no": "ACC-1033", "balance": 12000},
    {"account_no": "ACC-1034", "balance": 60000}
]

result = group_accounts(accounts)

for category, values in result.items():
    print(category, values)
```

### Q35. Write a decorator that logs checkout calls for banking case 1; use sample reference `ACC-1035` and explain the result.

**Hints**

1. First express the input and return value of one pure function.
2. Use `filter` for selection, `map` for transformation, and `reduce` only when a clearer built-in is not available.
3. Test the pipeline with an empty collection and avoid mutating account records in place.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
from functools import wraps


def log_checkout(func):
    @wraps(func)
    def wrapper(account_no, amount):
        print(f"Checkout started for {account_no}")

        result = func(account_no, amount)

        print(f"Checkout completed for {account_no}")

        return result

    return wrapper


@log_checkout
def checkout(account_no, amount):
    return f"Payment of {amount:.2f} processed"


result = checkout("ACC-1035", 2500)

print(result)
```
